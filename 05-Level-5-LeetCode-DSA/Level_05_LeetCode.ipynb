### 84. Largest Rectangle in Histogram (HARD)

**Category:** `NOT DONE`  
**Pattern:** LeetCode DSA Problem Solution



```python
#84. Largest Rectangle in Histogram (HARD)

def largestRectangle(heights):
    stack=[]
    max_area=0
    heights.append(0)
    
    for i in range(len(heights)):
        while stack and heights[stack[-1]]>heights[i]:
            h=heights[stack.pop()]
            
            if stack:
                width=i-stack[-1]-1
            else:
                width=i
            max_area=max(max_area,h*width)
        stack.append(i)
        
    return max_area

n=int(input())
heights=list(map(int,input().split()))
print(largestRectangle(heights))
```

### 496. Next Greater Element I (EASY)

**Category:** `NOT DONE`  
**Pattern:** LeetCode DSA Problem Solution



```python
#496. Next Greater Element I (EASY)

def nextGreaterElement(num1,num2):
    stack=[]
    greater={}
    
    for num in num2:
        while stack and stack[-1]<num:
            greater[stack.pop()]=num
        stack.append(num)
        
    for num in stack:
        greater[num]=-1
        
    return [greater[num] for num in num1]


n1=int(input())
num1=list(map(int,input().split()))

n2=int(input())
num2=list(map(int,input().split()))
print(nextGreaterElement(num1,num2))
```

### Convert Postfix to Prefix (Not on LeetCode, but a dojo question)

**Category:** `NOT DONE`  
**Pattern:** LeetCode DSA Problem Solution



```python
#Convert Postfix to Prefix (Not on LeetCode, but a dojo question)

def postfix_to_prefix(exp):
    stack=[]
    operators=['+','-','/','*']
    
    tokens=exp.split()
    
    for token in tokens:
        
        if token not in operators:
            stack.append(token)
        else:
            if len(stack)<2:
                return "Invalid Postfix Expression"
            
            right=stack.pop()
            left=stack.pop()
            
            new_exp=token+" "+left+" "+right
            stack.append(new_exp)
            
    return stack[-1]

exp=input().strip()

print(postfix_to_prefix(exp))
```

### 416. Partition Equal Subset Sum(MEDIUM)

**Category:** `NOT DONE`  
**Pattern:** LeetCode DSA Problem Solution



```python
#416. Partition Equal Subset Sum(MEDIUM)

def canPartition(nums):
    total=sum(nums)
    if total%2:
        return False
    
    target=total//2
    
    dp=[False]*(target+1)
    
    dp[0]=True
    
    for num in nums:
        for s in  range(target,num-1,-1):
            dp[s]=dp[s] or dp[s-num]
    return dp[target]

nums=list(map(int,input().split()))
canPartition(nums)
```

### 468. Validate IP Address(MEDIUM)

**Category:** `NOT DONE`  
**Pattern:** LeetCode DSA Problem Solution



```python
#468. Validate IP Address(MEDIUM)

def validIPaddress(queryIP):
    if queryIP.count('.')==3:
        parts=queryIP.split('.')
        
        for part in parts:
            if len(part)==0:
                return "Neither"
            
            if len(part)>1 and part[0]=='0':
                return 'Neither'
            
            if not part.isdigit():
                return 'Neither'
            
            if not (0<=int(part)<=255):
                return 'Neither'
        return 'IPv4'
    
    if queryIP.count(':')==7:
        parts=queryIP.split(':')
        
        hexa='0123456789abcdefABCDEF'
        
        for part in parts:
            if len(part)==0 or len(part)>4:
                return 'Neither'
            
            for ch in part:
                if not ch in hexa:
                    return 'Neither'
        return 'IPv6'
    return 'Neither'

queryIP=input()
validIPaddress(queryIP)
```

### 1040. Moving Stones Until Consecutive II(MEDIUM)

**Category:** `Sliding Window`  
**Pattern:** LeetCode DSA Problem Solution



```python
#1040. Moving Stones Until Consecutive II(MEDIUM)

def MovingStones(stones):
    stones.sort()
    n=len(stones)
    max_stones=max(stones[-1]-stones[1],stones[-2]-stones[0])-(n-2)
    
    min_stones=n
    left=0
    for right in range(n):
        while stones[right]-stones[left]+1>n:
            left+=1
        k=right-left+1
        if k==n and stones[right]-stones[left]+1==n:
            min_stones=min(min_stones,2)
        else:
            min_stones=min(min_stones,n-k)
    return [min_stones,max_stones]

stones=list(map(int,input().split()))
MovingStones(stones)
```

### 567. Permutation in String(MEDIUM)

**Category:** `Sliding Window`  
**Pattern:** LeetCode DSA Problem Solution



```python
#567. Permutation in String(MEDIUM)

from collections import Counter

def PermuteationString(s1,s2):
    if len(s1)>len(s2):
        return False
    
    s1_count=Counter(s1)
    window_count=Counter()
    
    left=0
    for right in range(len(s2)):
        window_count[s2[right]]+=1
        
        if right-left+1>len(s1):
            window_count[s2[left]]-=1
            if window_count[s2[left]]==0:
                del  window_count[s2[left]]
            left+=1
        
        if s1_count==window_count:
            return True
    return False

s1=input()
s2=input()
PermuteationString(s1,s2)
```

```python
#904. Fruit Into Baskets(MEDIUM)

def FruitsIntoBaskets(fruits):
    left=0
    ans=0
    basket={}
    
    for right in range(len(fruits)):
        basket[fruits[right]]=basket.get(fruits[right],0)+1
        
        if len(basket)>2:
            basket[fruits[left]]-=1
            if basket[fruits[left]]==0:
                del basket[fruits[left]]
            left+=1
        ans=max(ans,right-left+1)
    return ans

fruits=list(map(int,input().split()))
FruitsIntoBaskets(fruits)
```

```python
#1004. Max Consecutive Ones III(MEDIUM)

def MaxOnes(nums,k):
    zero=0
    left=0
    ans=0
    for right in range(len(nums)):
        if nums[right]==0:
            zero+=1
        
        while zero>k:
            if nums[left]==0:
                zero-=1
            left+=1
        ans=(max(ans,right-left+1))
    return ans

nums=list(map(int,input().split()))
k=int(input())
MaxOnes(nums,k)
```

### 2024. Maximize the Confusion of an Exam(MEDIUM)

**Category:** `Sliding Window`  
**Pattern:** LeetCode DSA Problem Solution



```python
#2024. Maximize the Confusion of an Exam(MEDIUM)

def MaxCofussion(answers,k):
    def sliding(char):
        left=0
        count=0
        ans=0
        for right in range(len(answers)):
            if answers[right]==char:
                count+=1
                
            while count>k:
                if answers[left]==char:
                    count-=1
                left+=1
            ans=max(ans,right-left+1)
        return ans
    return max(sliding('T'),sliding('F'))

answer=input()
k=int(input())
MaxCofussion(answer,k)
```

### 424. Longest Repeating Character Replacement(MEDIUM)

**Category:** `Sliding Window`  
**Pattern:** LeetCode DSA Problem Solution



```python
#424. Longest Repeating Character Replacement(MEDIUM)

def characterReplacement(s,k):
    left=0
    max_freq=0
    count={}
    ans=0
    
    for right in range(len(s)):
        count[s[right]]=count.get(s[right],0)+1
        
        max_freq=max(max_freq,count[s[right]])
        
        if (right-left+1)-max_freq>k:
            count[s[left]]-=1
            left+=1
        
        ans=max(ans,right-left+1)
    return ans 

s=input()
k=int(input())
characterReplacement(s,k)
```

```python
#674. Longest Continuous Increasing Subsequence(EASY)

def LICS(nums):
    if not nums:
        return 0
    
    left=0
    ans=1
    
    for right in range(1,len(nums)):
        if nums[right]<=nums[right-1]:
            left=right
            
        ans=max(ans,right-left+1)
    return ans

nums=list(map(int,input().split()))
LICS(nums)
```

**Category:** `Two Pointers`  
**Pattern:** LeetCode DSA Problem Solution



```python
#1679. Max Number of K-Sum Pairs(MEDIUM)

def MaxOperations(nums,k):
    nums.sort()
    n=len(nums)
    left=0
    right=n-1
    count=0
    
    while left<right:
        s=nums[left]+nums[right]
        if s==k:
            count+=1
            left+=1
            right-=1
        elif s<k:
            left+=1
        else:
            right-=1
    return count

nums=list(map(int,input().split()))
k=int(input())
MaxOperations(nums,k)
```

### 795. Number of Subarrays with Bounded Maximum(MEDIUM)

**Category:** `Two Pointers`  
**Pattern:** LeetCode DSA Problem Solution



```python
#795. Number of Subarrays with Bounded Maximum(MEDIUM)

def SubarrayMax(nums,left,right):
    def f(x):
        run=0
        count=0
        for val in nums:
            if val>x:
                run=0
            else:
                run+=1
            count+=run
        return count
    return f(right)-f(left-1)

nums=list(map(int,input().split()))
left=int(input())
right=int(input())
SubarrayMax(nums,left,right)
```

```python
#825. Friends Of Appropriate Ages(MEDIUM)

def FriendAges(ages):
    ages.sort()
    right,left,count=0,0,0
    n=len(ages)
    for i in range(n):
        if ages[i]<15:
            continue
        
        while ages[left]<=0.5*ages[i]+7:
            left+=1
        
        while right<n and ages[right]<=ages[i]:
            right+=1
        count+=right-left-1
    return count

ages=list(map(int,input().split()))
FriendAges(ages)
```

### 838. Push Dominoes(MEDIUM)

**Category:** `Two Pointers`  
**Pattern:** LeetCode DSA Problem Solution



```python
#838. Push Dominoes(MEDIUM)

def PushDominoes(dominoes):
    domi='L'+dominoes+'R'
    n=len(domi)
    left=0
    res=[]
    
    for right in range(1,n):
        if domi[right]=='.':
            continue
        
        mid=right-left-1
        
        if left>0:
            res.append(domi[left])
        
        if domi[left]==domi[right]:
            res.append(domi[left]*mid)
        elif domi[left]=='R' and domi[right]=='L':
            res.append('R'*(mid//2))
            if mid%2:
                res.append('.')
            res.append('L'*(mid//2))
        else:
            res.append("."*mid)
        left=right
    return "".join(res)

dominoes=input()
PushDominoes(dominoes)
```

```python
#2491. Divide Players Into Teams of Equal Skill(MEDIUM)

def dividePlayers(skill):
    skill.sort()
    n=len(skill)
    left=0
    right=n-1
    chem=0
    target=skill[left]+skill[right]
    
    while left<right:
        if skill[left]+skill[right]!=target:
            return -1
        
        chem+=skill[left]*skill[right]
        
        left+=1
        right-=1
    return chem

skill=list(map(int,input().split()))
dividePlayers(skill)
```

### 18. 4Sum(MEDIUM)

**Category:** `Two Pointers`  
**Pattern:** LeetCode DSA Problem Solution



```python
#18. 4Sum(MEDIUM)

def FourSum(nums,target):
    nums.sort()
    n=len(nums)
    result=[]
    
    for i in range(n-3):
        if i>0 and nums[i]==nums[i-1]:
            continue
        
        for j in range(i+1,n-2):
            if j>i+1 and nums[j]==nums[j-1]:
                continue
            
            left=j+1
            right=n-1
            while left<right:
                total=nums[i]+nums[j]+nums[left]+nums[right]
                
                if total==target:
                    result.append([nums[i],nums[j],nums[left],nums[right]])
                    while left<right and nums[left]==nums[left+1]:
                        left+=1
                    while left<right and nums[right]==nums[right-1]:
                        right-=1
                    left+=1
                    right-=1
                elif total<target:
                    left+=1
                else:
                    right-=1
    return result

nums=list(map(int,input().split()))
target=int(input())
FourSum(nums,target)
```

```python
#300. Longest Increasing Subsequence(MEDIUM)

import bisect

def LIS(nums):
    tails=[]
        
    for num in nums:
        idx=bisect.bisect_left(tails,num)
            
        if idx==len(tails):
            tails.append(num)
        else:
            tails[idx]=num
    return len(tails)
    
nums=list(map(int,input().split()))
LIS(nums)
```

### 875. Koko Eating Bananas(MEDIUM)

**Category:** `Binary Search`  
**Pattern:** LeetCode DSA Problem Solution



```python
#875. Koko Eating Bananas(MEDIUM)

def KokoBananas(piles,h):
    left=1
    right=max(piles)
    
    
    while right>left:
        mid=(right+left)//2
        hours=0
        for pile in piles:
            hours+=(pile+mid-1)//mid
            
        if hours<=h:
            right=mid
        else:
            left=mid+1
    return left

piles=list(map(int,input().split()))
h=int(input())
KokoBananas(piles,h)
```

### 1802. Maximum Value at a Given Index in a Bounded Array(MEDIUM)

**Category:** `Binary Search`  
**Pattern:** LeetCode DSA Problem Solution



```python
#1802. Maximum Value at a Given Index in a Bounded Array(MEDIUM)

def maxValue(n,index,maxSum):
    
    def cals(val,length):
        if val>length:
            big=val-1
            small=val-length
            return (big+small)*length//2
        
        k=val-1
        return k*(k+1)//2+(length-k)
    
    left=1
    right=maxSum
    while right>left:
        mid=(left+right+1)//2
        
        total=(
            mid+cals(mid,index)+cals(mid,n-index-1)
        )
        
        if total<=maxSum:
            left=mid
        else:
            right=mid-1
            
    return left

n=int(input())
index=int(input())
maxSum=int(input())
maxValue(n,index,maxSum)
```

```python
#2300. Successful Pairs of Spells and Potions(MEDIUM)

import bisect

def successfulPairs(spells,potion,success):
    potion.sort()
    n=len(potion)
    res=[]
    
    for spell in spells:
        target=(success+spell-1)//spell
        idx=bisect.bisect_left(potion,target)
        res.append(n-idx)
    return res

spells=list(map(int,input().split()))
potions=list(map(int,input().split()))
success=int(input())
successfulPairs(spells,potions,success)
```

### 2563. Count the Number of Fair Pairs(MEDIUM)

**Category:** `Binary Search`  
**Pattern:** LeetCode DSA Problem Solution



```python
#2563. Count the Number of Fair Pairs(MEDIUM)

import bisect

def FairPairs(nums,lower,upper):
    nums.sort()
    n=len(nums)
    ans=0
    
    for i in range(n):
        left=bisect.bisect_left(nums,lower-nums[i],i+1)
        right=bisect.bisect_right(nums,upper-nums[i],i+1)
        ans+=right-left
    return ans

nums=list(map(int,input().split()))
lower=int(input())
upper=int(input())
FairPairs(nums,lower,upper)
```

```python
#402. Remove K Digits(MEDIUM)

def RemoveKDigits(nums,k):
    stack=[]
    
    for num in nums:
        while stack and k>0 and stack[-1]>num:
            stack.pop()
            k-=1
        stack.append(num)
    while k>0:
        stack.pop()
        k-=1
    ans=''.join(stack).lstrip('0')
    return ans if ans else "0"

nums=input()
k=int(input())
RemoveKDigits(nums,k)
```

### 948. Bag of Tokens(MEDIUM)

**Category:** `Greedy & Stack`  
**Pattern:** LeetCode DSA Problem Solution



```python
#948. Bag of Tokens(MEDIUM)

def bagOfTokens(tokens,power):
    tokens.sort()
    
    score=0
    max_score=0
    left=0
    right=len(tokens)-1
    
    while left<=right:
        if power>=tokens[left]:
            power-=tokens[left]
            score+=1
            left+=1
            max_score=max(max_score,score)
        elif score>0:
            power+=tokens[right]
            right-=1
            score-=1
        else:
            break
    return max_score

tokens=list(map(int,input().split()))
power=int(input())
bagOfTokens(tokens,power)
```

```python
#1921. Eliminate Maximum Number of Monsters(MEDIUM)

def eleminateMax(dist,speed):
    n=len(dist)
    
    arrival=sorted(d/s for d,s in zip(dist,speed))
    
    for i,time in enumerate(arrival):
        if time<=i:
            return i
    return n

dist=list(map(int,input().split()))
speed=list(map(int,input().split()))
eleminateMax(dist,speed)
```

### 51. N-Queens(HARD)

**Category:** `Backtracking`  
**Pattern:** LeetCode DSA Problem Solution



```python
#51. N-Queens(HARD)

def NQueens(n):
    result=[]
    board=[['.']*n for _ in range(n)]
    def backtrack(row,cols,diag1,diag2,board):
        if row==n:
            result.append([''.join(r)for r in board])
            return
        
        for col in range(n):
            if col in cols or (row-col) in diag1 or (row+col) in diag2:
                continue
            
            board[row][col]='Q'
            cols.add(col)
            diag1.add(row-col)
            diag2.add(row+col)
            
            backtrack(row+1,cols,diag1,diag2,board)
            
            board[row][col]='.'
            cols.remove(col)
            diag1.remove(row-col)
            diag2.remove(row+col)
            
    backtrack(0,set(),set(),set(),board)
    return result

n=int(input())
NQueens(n)
```

### 1239. Maximum Length of a Concatenated String with Unique Characters(MEDIUM)

**Category:** `Backtracking`  
**Pattern:** LeetCode DSA Problem Solution



```python
#1239. Maximum Length of a Concatenated String with Unique Characters(MEDIUM)

def MaxLength(arr):
    result=0
    def backtrack(start,current):
        nonlocal result
        result=max(result,len(current))
        
        for i in range(start,len(arr)):
            if len(arr[i])!=len(set(arr[i])):
                continue
            
            if set(current) & set(arr[i]):
                continue
            
            backtrack(i+1,current+arr[i])
    backtrack(0,"")
    return result

arr=input().split()
MaxLength(arr)
```

### 17. Letter Combinations of a Phone Number(MEDIUM)

**Category:** `Backtracking`  
**Pattern:** LeetCode DSA Problem Solution



```python
#17. Letter Combinations of a Phone Number(MEDIUM)

def LettersCombinations(digit):
    if not digit:
        return []
    
    phone={
        '2':'abc',
        '3':'def',
        '4':'ghi',
        '5':'jkl',
        '6':'mno',
        '7':'pqrs',
        '8':'tuv',
        '9':'xyz'
    }
    
    result=[]
    
    def backtrack(index,path):
        if index==len(digit):
            result.append(path)
            return

        for ch in phone[digit[index]]:
            backtrack(index+1,path+ch)
            
    backtrack(0,'')
    return result

digit=input()
LettersCombinations(digit)
```

### 22. Generate Parentheses(MEDIUM)

**Category:** `Backtracking`  
**Pattern:** LeetCode DSA Problem Solution



```python
#22. Generate Parentheses(MEDIUM)

def generateParenthesis(n):
    result=[]
    
    def backtrack(path,open,close):
        if len(path)==2*n:
            result.append(path)
            return 
        
        if open<n:
            backtrack(path+'(',open+1,close)
        
        if close<open:
            backtrack(path+')',open,close+1)
    backtrack('',0,0)
    return result

n=int(input())
generateParenthesis(n)
```

### 131. Palindrome Partitioning(MEDIUM)

**Pattern:** LeetCode DSA Problem Solution



```python
#131. Palindrome Partitioning(MEDIUM)

def partition(s):
    result=[]
    
    def isPanlindrome(ch):
        return ch==ch[::-1]
    
    def backtrack(start,path):
        if len(s)==start:
            result.append(path[:])
            return
        
        for i in range(start,len(s)):
            string=s[start:i+1]
            
            if isPanlindrome(string):
                path.append(string)
                backtrack(i+1,path)
                path.pop()
    backtrack(0,[])
    return result

s=input()
partition(s)
```

### 1219. Path with Maximum Gold(MEDIUM)

**Category:** `Backtracking`  
**Pattern:** LeetCode DSA Problem Solution



```python
#1219. Path with Maximum Gold(MEDIUM)

def getMaxGold(grid):
    rows,cols=len(grid),len(grid[0])
    max_gold=0
    
    def dfs(r,c):
        if r<0 or c<0 or r>=rows or c>=cols or grid[r][c]==0:
            return 0
        
        gold=grid[r][c]
        grid[r][c]=0
        
        up=dfs(r-1,c)
        down=dfs(r+1,c)
        left=dfs(r,c-1)
        right=dfs(r,c+1)
        
        grid[r][c]=gold
        return gold+max(up,down,left,right)
    
    for r in range(rows):
        for c in range(cols):
            if grid[r][c]!=0:
                max_gold=max(max_gold,dfs(r,c))
    return max_gold

grid=[]
for i in range(3):
    row = list(map(int, input().split()))
    grid.append(row)
getMaxGold(grid)
```

### 46. Permutations(MEDIUM

**Pattern:** LeetCode DSA Problem Solution



```python
#46. Permutations(MEDIUM

def permutation(nums):
    res=[]
    def backtrack(start):
        if start==len(nums):
            res.append(nums[:])
            
        for i in range(start,len(nums)):
            nums[start],nums[i]=nums[i],nums[start]
            backtrack(start+1)
            nums[i],nums[start]=nums[start],nums[i]
    backtrack(0)
    return res

nums=list(map(int,input().split()))
permutation(nums)
```

```python
#491. Non-decreasing Subsequences(MEDIUM)

def findSubsequence(nums):
    result=[]
    def backtrack(start,path):
        if len(path)>=2:
            result.append(path[:])
            
        used=set()
        for i in range(start,len(nums)):
            if nums[i] in used:
                continue
            
            if path and nums[i]<path[-1]:
                continue
            
            used.add(nums[i])
            path.append(nums[i])
            backtrack(i+1,path)
            path.pop()
    backtrack(0,[])
    return result

nums=list(map(int,input().split()))
findSubsequence(nums)
```

### 526. Beautiful Arrangement(MEDIUM)

**Category:** `Backtracking`  
**Pattern:** LeetCode DSA Problem Solution



```python
#526. Beautiful Arrangement(MEDIUM)

def BeautifulArr(n):
    count=0
    visited=set()
    
    def backtrack(pos):
        nonlocal count
        if pos>n:
            count+=1
            return
        
        for num in range(1,n+1):
            if num in visited:
                continue
            
            if num%pos==0 or pos%num==0:
                visited.add(num)
                backtrack(pos+1)
                visited.remove(num)
    backtrack(1)
    return count

n=int(input())
BeautifulArr(n)
```

```python
#2597. The Number of Beautiful Subsets(MEDIUM)

def BeautifulSubsets(nums,k):
    count=0
    freq={}
    nums.sort()
    
    def backtrack(start):
        nonlocal count
        if start>0:
            count+=1
            
        
        for i in range(start,len(nums)):
            if freq.get(nums[i]-k,0)>0:
                continue
            
            freq[nums[i]]=freq.get(nums[i],0)+1
            backtrack(i+1)
            freq[nums[i]]-=1
    backtrack(0)
    return count

nums=list(map(int,input().split()))
k=int(input())
BeautifulSubsets(nums,k)
```

```python
#3211. Generate Binary Strings Without Adjacent Zeros(MEDIUM)

def validString(n):
    result=[]
    def backtrack(path):
        if len(path)==n:
            result.append(path)
            return
        
        backtrack(path+'1')
        
        if not path or path[-1]!='0':
            backtrack(path+'0')
            
    backtrack('')
    return result

n=int(input())
validString(n)
```

### 93. Restore IP Addresses(MEDIUM)

**Category:** `Backtracking`  
**Pattern:** LeetCode DSA Problem Solution



```python
#93. Restore IP Addresses(MEDIUM)

def restoreIPaddress(s):
    result=[]
    
    def isValid(part):
        if len(part)>1 and part[0]=='0':
            return False
        
        if int(part)>255:
            return False
        return True
    
    def backtrack(start,parts):
        if len(parts)==4 and start==len(s):
            result.append('.'.join(parts))
            return
        
        if len(parts)==4 or start==len(s):
            return
        
        for right in range(1,4):
            if start+right>len(s):
                break
            
            part=s[start:start+right]
            if isValid(part):
                parts.append(part)
                backtrack(start+right,parts)
                parts.pop()
    backtrack(0,[])
    return result

s=input()
restoreIPaddress(s)
```
