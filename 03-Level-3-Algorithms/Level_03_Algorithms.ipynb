### Problem 01: Second Largest Element in Array

**Problem Statement:**
Given an array of integers, find the second largest distinct element. If no second largest exists, return the maximum element.

**Input Format:**

The second largest element.

**Example:**
```text
Input:
5

10
```

### Problem 02: Forward Prefix Sum Array

**Problem Statement:**
Compute the prefix sum array from left to right where each element is the sum of all elements up to that index.

### Problem 04: Bubble Sort Algorithm

**Problem Statement:**
Implement the Bubble Sort algorithm to sort an array in ascending order by repeatedly swapping adjacent inverted elements.

**Input Format:**
Line 1: Length `n`.
Line 2: `n` space-separated integers.

**Output Format:**
Space-separated sorted array.

**Example:**
```text
Input:
5
5 1 4 2 8
Output:
1 2 4 5 8
```



````python
#BUBBLE SORT

n=int(input())
lst=list(map(int,input().split()))

for i in range(n-1):
    for j in range(n-1-i):
        if lst[j]>lst[j+1]:
            lst[j],lst[j+1]=lst[j+1],lst[j]
            
print(*lst)
````

### Problem 05: Insertion Sort Algorithm

**Problem Statement:**
Implement the Insertion Sort algorithm to sort an array in ascending order by building the sorted array one item at a time.

**Input Format:**
Line 1: Length `n`.
Line 2: `n` space-separated integers.

**Output Format:**
Space-separated sorted array.

**Example:**
```text
Input:
5
12 11 13 5 6
Output:
5 6 11 12 13
```



````python
#INSERTION SORT

n=int(input())
lst=list(map(int,input().split()))

for i in range(1,n):
    key=lst[i]
    j=i-1
    
    while j>=0 and lst[j]>key:
        lst[j+1]=lst[j]
        j-=1
    lst[j+1]=key
print(*lst)
````

### Problem 06: Selection Sort Algorithm

**Problem Statement:**
Implement the Selection Sort algorithm to sort an array in ascending order by repeatedly finding the minimum element from the unsorted part.

**Input Format:**
Line 1: Length `n`.
Line 2: `n` space-separated integers.

**Output Format:**
Space-separated sorted array.

**Example:**
```text
Input:
5
64 25 12 22 11
Output:
11 12 22 25 64
```



````python
#SELECTION SORT

n=int(input())
lst=list(map(int,input().split()))

for i in range(n):
    min_index=i
    for j in range(i+1,n):
        if lst[j]<lst[min_index]:
            min_index=j
    lst[i],lst[min_index]=lst[min_index],lst[i]
            
print(*lst)
````

### Problem 08: Fast Power Exponentiation Pow(x, n)

**Problem Statement:**
Implement $x^n$ using recursive divide-and-conquer binary exponentiation in $O(\log n)$ time, handling positive, negative, and zero exponents.

**Input Format:**
Two space-separated floats `x` and `n`.

**Output Format:**
Result of $x^n$.

**Example:**
```text
Input: 2.0 10
Output: 1024.0
```



````python
def power(x,n):
    if n==0:
        return 1
    elif n<0:
        return 1/power(x,-n)
    else:
        half=power(x,n//2)
        if n%2==0:
            return half*half
        else:
            return x*half*half
        
x,n=map(float,input().split())
n=int(n)

result=power(x,n)
print(f"{result:.2f}")
````

### Problem 09: Character Frequency Analysis (High & Low Extremes)

**Problem Statement:**
Count character frequencies in a string and identify characters with the highest and lowest frequencies.

**Input Format:**
A string `n`.

**Output Format:**
Lists of highest and lowest frequency characters.

**Example:**
```text
Input: abracadabra
Output: Highest: ['a'], Lowest: ['c', 'd']
```



````python
n=input()
freq={}

for ch in n:
    if ch in freq:
        freq[ch]+=1
    else:
        freq[ch]=1

high_count=max(freq.values())
low_count=min(freq.values())

high_letter=sorted([ch for ch,f in freq.items() if f==high_count])
low_letter=sorted([ch for ch,f in freq.items() if f==low_count])

print("Highest Frequency:"," ".join(high_letter),high_count)
print("Lowest Frequency:"," ".join(low_letter),low_count)
````

### Problem 11: Manual String Whitespace Trimming

**Problem Statement:**
Strip leading and trailing spaces from a string manually using index pointers without relying on built-in `.strip()`.

**Input Format:**
A string with possible leading/trailing whitespace.

**Output Format:**
The trimmed string.

**Example:**
```text
Input: "   hello world   "
Output: "hello world"
```



````python
string=input()

start=0
while start<len(string) and string[start]==" ":
    start+=1
    
end=len(string)-1
while end>=0 and string[end]==" ":
    end-=1
    
if start<=end:
    result=string[start:end+1]
else:
    result=""
    
print(result)
````

### Problem 15: Remove First Occurrence of Substring (Manual Matching)

**Problem Statement:**
Find and remove the first occurrence of substring `w` using manual index slicing and matching loops without built-in replacement.

**Input Format:**
Line 1: Text `n`.
Line 2: Substring `w`.

**Output Format:**
Resulting string after removal.

**Example:**
```text
Input:
abcdef
cd
Output:
abef
```



````python
n=input()
w=input()

pos=-1
for i in range(len(n)-len(w)+1):
    match=True
    for j in range(len(w)):
        if n[i+j]!=w[j]:
            match=False
            break
    if match:
        pos=i
        break
    
if pos!=-1:
    n=n[:pos]+n[pos+len(w):]
print(n)
````
