### Problem 02: Sum Square Difference (Project Euler #6)

**Problem Statement:**
Find the difference between the square of the sum of the first $n$ natural numbers and the sum of the squares of the first $n$ natural numbers: $(1 + 2 + ... + n)^2 - (1^2 + 2^2 + ... + n^2)$.

**Input Format:**
A single integer `n`.

**Output Format:**
The difference between the square of sum and sum of squares.

**Example:**
```text
Input: 10
Output: 2640
```



````python
n=int(input())

s1=0
s2=0

for i in range(1,n+1):
    s1+=int(math.pow(i,2))
    s2+=i
    
sq_of_sum=math.pow(s2,2)

print(int(sq_of_sum-s1))
````

### Problem 03: Sum of Digits

**Problem Statement:**
Given an integer $n$, compute the sum of all its digits.

**Input Format:**
A single integer `n`.

**Output Format:**
The sum of digits of `n`.

**Example:**
```text
Input: 1234
Output: 10
```



````python
n=int(input())

digit=0

while(n>0):
    digit+=n%10
    n//=10
print(digit)
````

### Problem 05: Build Array from Permutation (LeetCode 1920)

**Problem Statement:**
Given a zero-based permutation array `nums` of length $n$, construct an array `ans` of the same length where `ans[i] = nums[nums[i]]` for each $0 \le i < n$.

**Input Format:**
Line 1: Length `n`.
Line 2: `n` space-separated integers.

**Output Format:**
Space-separated values of array `ans`.

**Example:**
```text
Input:
6
0 2 1 5 3 4
Output:
0 1 2 4 5 3
```



````python
n=int(input())
nums=list(map(int,input().split()))
ans=[]
for i in range(n):
    ans.append(nums[nums[i]])
print(*ans)
````

### Problem 06: Final Value of Variable After Operations (LeetCode 2011)

**Problem Statement:**
Starting from $X = 0$, process a sequence of operations (`++X`, `X++`, `--X`, `X--`) and print the final value of $X$.

**Input Format:**
Line 1: Number of operations `n`.
Line 2: Space-separated operation tokens.

**Output Format:**
Final integer value of `X`.

**Example:**
```text
Input:
3
--X X++ X++
Output:
1
```



````python
n=int(input())
oper=input().split()
x=0

for op in oper:
    if op in ['x++','++x']:
        x+=1
    else:
        x-=1
x
````

### Problem 07: Running Sum of 1D Array (LeetCode 1480)

**Problem Statement:**
Given an array `nums`, compute the running sum where `runningSum[i] = sum(nums[0]...nums[i])`.

**Input Format:**
Line 1: Array length `n`.
Line 2: `n` space-separated integers.

**Output Format:**
Space-separated running sums.

**Example:**
```text
Input:
4
1 2 3 4
Output:
1 3 6 10
```



````python
n=int(input())
nums=list(map(int,input().split()))
run=[]
curr=0
for num in nums:
    curr+=num
    run.append(curr)
print(*run)
````

### Problem 08: Check for Duplicate Elements (Contains Duplicate)

**Problem Statement:**
Check if an array contains duplicate elements by comparing all pairs.

**Input Format:**
Line 1: Integer `n`.
Line 2: `n` space-separated integers.

**Output Format:**
`Yes` if any pair matches, else `No`.

**Example:**
```text
Input:
4
1 2 3 1
Output:
Yes
```



````python
n=int(input())
lst=list(map(int,input().split()))

for i in range(0,n):
    for j in range(i+1,n):
        if lst[i]==lst[j]:
            print("Yes")
        else:
            print("No")
````

### Problem 09: Tiered Water / Utility Bill Calculation

**Problem Statement:**
Calculate the utility bill for $n$ units consumed based on consumption tiers:
- 1 to 10 units: $2.00 / unit
- 11 to 30 units: $3.50 / unit
- 31 to 50 units: $5.00 / unit
- Units above 50: $8.00 / unit

**Input Format:**
Integer `n` representing consumption units.

**Output Format:**
Total bill formatted to 2 decimal places.

**Example:**
```text
Input: 45
Output: 165.00
```



````python
n=int(input())

cost=0.0

if(n>0):
    basic=min(n,10)
    cost+=basic*2.00

if(n>10):
    standard=min(n-10,20)
    cost+=standard*3.50
    
if(n>30):
    heavy=min(n-30,20)
    cost+=heavy*5.00
    
if(n>50):
    excessive=n-50
    cost+=excessive*8.00
    
print(f"{cost:.2f}")
````

### Problem 10: Body Mass Index (BMI) & Category Classifier

**Problem Statement:**
Calculate BMI ($w / h^2$) and categorize:
- $BMI < 18.5$: Underweight
- $18.5 \le BMI < 24.9$: Normal weight
- $25.0 \le BMI < 29.9$: Overweight
- $BMI \ge 30.0$: Obesity

**Input Format:**
Line 1: Weight in kg (`w`).
Line 2: Height in meters (`h`).

**Output Format:**
BMI value (1 decimal place) and category name.

**Example:**
```text
Input:
70
1.75
Output:
BMI:22.9
Normal weight
```



````python
w=float(input())
h=float(input())

BMI=w/(h*h)
print(f"BMI:{BMI:.1f}")
if BMI<18.5:
    print("Underweight")
elif (18.5<=BMI<24.9):
    print("Normal weight")
elif(25.0<=BMI<29.9):
    print("Overweight")
elif(BMI>=30.0):
    print("Obesity")
````

### Problem 11: Shopping Discount Slabs and Final Bill

**Problem Statement:**
Calculate discount percentage and final bill:
- Amount $\ge 500$: 20% discount
- $300 \le Amount \le 499$: 15% discount
- $200 \le Amount \le 299$: 10% discount
- Amount $< 200$: 0% discount

**Input Format:**
Float representing initial total bill.

**Output Format:**
Discount percentage and final amount formatted to 2 decimal places.

**Example:**
```text
Input: 400
Output:
Discount: 15%
Final Amount: 340.00
```



````python
n=float(input())

if(n>=500):
    print("Discount: 20%")
    d=n*0.20
    n-=d
    print(f"Final Amount: {n:.2f}")
elif(300<=n<=499):
    print("Discount: 15%")
    d=n*0.15
    n-=d
    print(f"Final Amount: {n:.2f}")    
elif(200<=n<=299):
    print("Discount: 10%")
    d=n*0.10
    n-=d
    print(f"Final Amount: {n:.2f}")
elif(n<200):
    print("Discount: 0%")
    print(f"Final Amount: {n:.2f}")
````
