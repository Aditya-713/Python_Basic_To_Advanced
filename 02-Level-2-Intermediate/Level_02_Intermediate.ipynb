### Problem 01: Count Prime Numbers in a List

**Problem Statement:**
Given a list of positive integers, count how many numbers in the list are prime numbers.

**Input Format:**
A space-separated list of integers.

**Output Format:**
Count of prime numbers.

**Example:**
```text
Input: 2 3 4 5 6 7 8 9 10
Output: 4
```



````python
a=list(map(int,input().split()))
count=0

for num in a:
    if num>1:
        is_prime=True
        for i in range(2,int(num**0.5)+1):
            if num%2==0:
                is_prime=False
                break
        if is_prime==True:
            count+=1
print(count)
````

### Problem 02: Filter Out 'b' and 'ac' from String

**Problem Statement:**
Given a string, remove all occurrences of the character `'b'` and the substring `'ac'` in a single pass.

**Input Format:**
A single string `s`.

**Output Format:**
The modified string without `'b'` and `'ac'`.

**Example:**
```text
Input: react
Output: ret
```



````python
s=input().lower()
result=[]
i=0

while i<len(s):
    if s[i]=='b':
        i+=1
    elif i+1<len(s) and s[i]=='a' and s[i+1]=='c':
        i+=2
    else:
        result.append(s[i])
        i+=1
print("".join(result))
````

### Problem 07: Maximum Product of K Consecutive Digits

**Problem Statement:**
Given a numeric string $n$ and an integer $k$, find the maximum product among all contiguous substrings of length $k$.

**Input Format:**
Line 1: Window size `k`.
Line 2: Numeric string `n`.

**Output Format:**
Maximum product of $k$ consecutive digits.

**Example:**
```text
Input:
3
3675356291
Output:
315
```



````python
k=int(input())
n=input()
max_product=0

length=len(n)
for i in range(length-k+1):
    product=1
    for j in range(i,i+k):
        product*=int(n[j])
    if product>max_product:
        max_product=product

print(max_product)
````

### Problem 11: Recursive Fibonacci Terms

**Problem Statement:**
Print the first $a+1$ terms (from index 0 to $a$) of the Fibonacci sequence using a recursive function.

**Input Format:**
An integer `a`.

**Output Format:**
Space-separated Fibonacci terms.

**Example:**
```text
Input: 6
Output: 0 1 1 2 3 5 8
```



````python
def fibo(n):
    if n<=1:
        return n
    return fibo(n-1)+fibo(n-2)

a=int(input())
for i in range(a+1):
    print(fibo(i),end=' ')
````

### Problem 12: Recursive Pentagonal Numbers

**Problem Statement:**
Compute and print the first $a$ pentagonal numbers using recursion: $P(1) = 1$, $P(n) = P(n-1) + (3n - 2)$.

**Example:**

### Problem 13: Armstrong (Narcissistic) Number Check

**Problem Statement:**
Determine if an integer $n$ is an Armstrong number (the sum of each digit raised to the power of total number of digits equals $n$).

**Input Format:**
An integer `n`.

**Output Format:**
`Armstrong` or `Not Armstrong`.

**Example:**
```text
Input: 153
Output: Armstrong
```



````python
n=int(input())
temp=n
a=0
st=len(str(n))

while temp>0:
    digit=temp%10
    a+=digit**st
    temp//=10
    
if a==n:
    print("Armstrong")
else:
    print("Not Armstrong")
````

### Problem 14: Perfect Number Check

**Problem Statement:**
Determine if a positive integer $n$ is a perfect number (equal to the sum of its proper divisors, excluding itself).

**Input Format:**
An integer `n`.

**Output Format:**
`Perfect` or `Not Perfect`.

**Example:**
```text
Input: 28
Output: Perfect
```



````python
n=int(input())
a=0

for i in range(1,int(n**0.5)+1):
    if n%i==0:
        a+=i
        if n!=1 and n!=n//i:
            a+=n//i
    
if a==n:
    print("Perfect")
else:
    print("Not Perfect")
````

### Problem 29: Spiral Order Matrix Traversal

**Problem Statement:**
Given an $n \times m$ matrix, traverse and collect all its elements in clockwise spiral order.

**Input Format:**
Line 1: Dimensions `n m`.
Next `n` lines: `m` space-separated integers.

**Output Format:**
List of elements in spiral order.

**Example:**
```text
Input:
3 3
1 2 3
4 5 6
7 8 9
Output:
[1, 2, 3, 6, 9, 8, 7, 4, 5]
```



````python
n,m=map(int,input().split())

matrix=[]

for i in range(n):
    row=list(map(int,input().split()))
    matrix.append(row)
    
result=[]

top,bottom,left,right=0,n-1,0,m-1

while top<=bottom and left<=right:
    for i in range(left,right+1):
        result.append(matrix[top][i])
    top+=1
    
    for i in range(top,bottom+1):
        result.append(matrix[i][right])
    right-=1
    
    if top<=bottom:
        for i in range(right,left-1,-1):
            result.append(matrix[bottom][i])
        bottom-=1
        
    if left<=right:
        for i in range(bottom,top-1,-1):
            result.append(matrix[i][left])
        left+=1
        
prime=0
for num in result:
    if num>1:
        is_prime=True 
        for i in range(2,int(num**0.5)+1):
            if num%i==0:
                is_prime=False
                break
        
        if is_prime:
            prime+=num

for row in matrix:            
    print(*row)
print(*result)
print(prime)
````

### Problem 30: Swap Min and Max Elements in Matrix

**Problem Statement:**
Find the global minimum and maximum values in a matrix and swap all occurrences of min with max and vice versa.

**Input Format:**
Line 1: Order `n`.
Next `n` lines: `n` space-separated integers.

**Output Format:**
Matrix with min and max elements swapped.

**Example:**
```text
Input:
2
1 9
5 2
Output:
9 1 
5 2 
```



````python
n=int(input())
m1=[]

for i in range(n):
    row=list(map(int,input().split()))
    m1.append(row)
    
s=min(min(row) for row in m1)
l=max(max(row) for row in m1)

for i in range(n):
    for j in range(n):
        if m1[i][j]==s:
            m1[i][j]=l
        elif m1[i][j]==l:
            m1[i][j]=s
            

for i in range(n):
    for j in range(i+1,n):
        m1[i][j],m1[j][i]=m1[j][i],m1[i][j]
        
for row in m1:
    row.reverse()
    
for row in m1:
    print(*row)
````

### Problem 31: Longest Common Prefix

**Problem Statement:**
Given a space-separated list of words, find the longest common prefix string shared among all the words.

**Input Format:**
A space-separated list of strings.

**Output Format:**
The longest common prefix string.

**Example:**
```text
Input: flower flow flight
Output: fl
```



````python
n=input().split()

n.sort()

s1=n[0]
s2=n[-1]

ans=''

i=0
while i<len(s1) and i<len(s2):
    if s1[i]==s2[i]:
        ans+=s1[i]
    else:
        break
    i+=1

print(f"{ans}")
````
