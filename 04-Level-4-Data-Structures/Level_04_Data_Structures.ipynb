### Problem 04: Maximum Subarray Sum (Kadane's Algorithm)

**Problem Statement:**
Find the maximum sum of any contiguous subarray within a given array of numbers using Kadane's Algorithm.

**Input Format:**
Array of integers.

**Output Format:**
Maximum subarray sum.

**Example:**
```text
Input: -2 1 -3 4 -1 2 1 -5 4
Output: 6 (Subarray [4, -1, 2, 1])
```



````python
#Maximum Subarray Sum

def max_subarray(lst):
    if len(lst)==0:
        return 0
    
    max_num=lst[0]
    curr_num=0
    
    for num in lst:
        curr_num+=num
        max_num=max(max_num,curr_num)
        
        if curr_num<0:
            curr_num=0
            
    return max_num

n=int(input())
arr=list(map(int,input().split()))
print(max_subarray(arr))
````

### Problem 05: Check if Any Rotation of String is a Palindrome

**Problem Statement:**
Determine if any cyclic rotation of a given string forms a valid palindrome.

**Input Format:**
A string `s`.

**Output Format:**
`True` if any rotation is a palindrome, else `False`.

**Example:**
```text
Input: aab
Output: True (Rotation 'aba')
```



````python
#Check if Any Rotation of a String is a Palindrome

def can_be_palindrome(s):
    n=len(s)
    
    def palindrome(t):
        return t==t[::-1]
    
    for i in range(n):
        rotated=s[i:]+s[:i]
        if palindrome(rotated):
            return True
        
    return False

a=input()
if can_be_palindrome(a):
    print("Yes, Rotated string is a palindrome")
else:
    print("No, Rotated string is not a palindrome")
````

### Problem 06: Check if Pair with Given Difference Exists

**Problem Statement:**
Given an array of integers and a difference $k$, check if there exists any pair $(i, j)$ such that $|arr[i] - arr[j]| = k$.

**Input Format:**
Array of integers and target difference `k`.

**Output Format:**
`True` or `False`.

**Example:**
```text
Input: arr = [1, 5, 3, 4, 2], k = 3
Output: True
```



````python
#Check if a Pair with Given Difference Exists

def pair_exist(arr,k):
    n=len(arr)
    i,j=0,1
    
    while i<n and j<n:
        if i!=j:
            diff=arr[j]-arr[i]
            
            if diff==k:
                return 1
            elif diff<k:
                j+=1
            else:
                i+=1
        else:
            j+=1
    return 0

a=int(input())
lst=list(map(int,input().split()))
k=int(input())

print(pair_exist(lst,k))
````

### Problem 07: Reverse a Singly Linked List

**Problem Statement:**
Reverse the pointer links of a singly linked list in-place and return the new head.

**Input Format:**
Linked list nodes `1 -> 2 -> 3 -> 4 -> 5`.

**Output Format:**
Reversed linked list `5 -> 4 -> 3 -> 2 -> 1`.

**Example:**
```text
Input: 1 2 3 4 5
Output: 5 4 3 2 1
```



````python
#Reversed Linked list

class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        
def reverse_list(head):
    prev=None
    curr=head
    
    while curr:
        nxt=curr.next
        curr.next=prev
        prev=curr
        curr=nxt
    return prev

def display(head):
    temp=head
    while temp:
        print(temp.data,end=' ')
        temp=temp.next
        
n=int(input())
if n==0:
    print("")
else:
    values=list(map(int,input().split()))
    head=Node(values[0])
    temp=head
    for v in values[1:]:
        temp.next=Node(v)
        temp=temp.next
    head=reverse_list(head)
    display(head)
````

### Problem 08: Find the Town Judge (Graph Indegree)

Judge ID or -1 if no judge exists.

**Example:**
```text
Input: n = 3, trust = [[1,3],[2,3]]

```



````python
#Town Judge

m=int(input())
n=int(input())

score=[0]*(n+1)

for _ in range(m):
    a,b=map(int,input().split())
    score[a]-=1
    score[b]+=1
    
judge=-1

for person in range(1,n+1):
    if score[person]==(n-1):
        judge=person
        break
    
print(judge)
````

### Problem 09: Middle Element of Linked List

**Problem Statement:**


### Problem 10: Rearrange Linked List by Odd and Even Positions

**Problem Statement:**
Group all odd-indexed nodes together followed by even-indexed nodes in a singly linked list.

**Input Format:**
Linked list nodes `1 -> 2 -> 3 -> 4 -> 5`.

**Output Format:**
`1 -> 3 -> 5 -> 2 -> 4`.

**Example:**

````python
#Rearrange a Linked List by Odd and Even Positions

class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        
def odd_even(head):
    if not head or not head.next:
        return head
    
    odd=head
    even=odd.next
    even_head=even
    
    while even and even.next:
        odd.next=even.next
        odd=odd.next
        
        even.next=odd.next
        even=even.next
        
    odd.next=even_head
    return head

def display(head):
    if not head:
        print("No Elements")
        return
    
    temp=head
    while temp:
        print(temp.data,end=' ')
        temp=temp.next
        
n=int(input())
if n==0:
    print("No elements")
else:
    arr=list(map(int,input().split()))
    
    head=Node(arr[0])
    temp=head
    for val in arr[1:]:
        temp.next=Node(val)
        temp=temp.next
        
order=odd_even(head)
display(order)
````

### Problem 11: Sort Linked List Using Merge Sort

**Problem Statement:**
Implement merge sort on a singly linked list with $O(n \log n)$ time complexity.

**Input Format:**
Unsorted linked list `4 -> 2 -> 1 -> 3`.

**Output Format:**
Sorted linked list `1 -> 2 -> 3 -> 4`.

**Example:**
```text
Input: 4 2 1 3
Output: 1 2 3 4
```



````python
#Sort a Singly Linked List Using Merge Sort

class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        
def middle(head):
    if not head:
        return head
    
    slow=head
    fast=head.next
    
    while fast and fast.next:
        slow=slow.next
        fast=fast.next.next
        
    return slow

def merge(left,right):
    if not left:
        return right
    if not right:
        return left
    
    if left.data<=right.data:
        result=left
        result.next=merge(left.next,right)
    else:
        result=right
        result.next=merge(left,right.next)
    
    return result

def merge_sort(head):
    if not head or not head.next:
        return head
    
    mid=middle(head)
    next_to_mid=mid.next
    mid.next=None
    
    left=merge_sort(head)
    right=merge_sort(next_to_mid)
    return merge(left,right)

def display(head):
    temp=head
    while temp:
        print(temp.data,end=' ')
        temp=temp.next
        

n=int(input())
arr=list(map(int,input().split()))

head=Node(arr[0])
temp=head
for val in arr[1:]:
    temp.next=Node(val)
    temp=temp.next
    
head=merge_sort(head)
display(head)
````

### Problem 12: Delete All Nodes with Value X

**Problem Statement:**
Given a linked list and a target value $x$, delete all nodes whose data equals $x$.

**Input Format:**
Linked list and target value `x`.

**Output Format:**
Modified linked list without nodes of value `x`.

**Example:**
```text
Input: 1 2 6 3 4 6, x = 6
Output: 1 2 3 4
```



````python
#Delete all nodes whose value is X

class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        
def delete_x(head,x):
    while head and head.data==x:
        head=head.next
    
    curr=head    
    while curr and curr.next:
        if curr.next.data==x:
            curr.next=curr.next.next
        else:
            curr=curr.next
        
    return head


def display(head):
    temp=head
    while temp:
        print(temp.data,end=' ') 
        temp=temp.next
        
n=int(input())
arr=list(map(int,input().split()))
x=int(input())

head=Node(arr[0])
temp=head
for val in arr[1:]:
    temp.next=Node(val)
    temp=temp.next
    
dele=delete_x(head,x)
display(dele)
````

### Problem 14: Design MinStack (O(1) getMin)

**Problem Statement:**
Design a stack class `MinStack` that supports `push`, `pop`, `top`, and retrieving the minimum element in $O(1)$ time.

**Input Format:**
Sequence of MinStack method calls.

**Output Format:**
Values returned by top and getMin.

**Example:**
```text
push(-2), push(0), push(-3), getMin() -> -3, pop(), top() -> 0, getMin() -> -2
```



````python
class MinStack:
    def __init__(self):
        self.stack=[]
        self.minstack=[]
        
    def push(self,val):
        self.stack.append(val)
        if not self.minstack:
            self.minstack.append(val)
        else:
            self.minstack.append(
                val if val<self.minstack[-1] else self.minstack[-1]
            )
        
    def pop(self):
        if not self.stack:
            return -1
        self.minstack.pop()
        return self.stack.pop()
    
    def getmin(self):
        if not self.minstack:
            return -1
        return self.minstack[-1]
    

q=int(input())
ms=MinStack()

for _ in range(q):
    arr=list(map(int,input().split()))
    
    if arr[0]==1:
        ms.push(arr[1])
    elif arr[0]==2:
        print(ms.pop())
    elif arr[0]==3:
        print(ms.getmin())
````

### Problem 16: Palindrome Linked List Check

**Problem Statement:**
Determine whether a singly linked list is a palindrome.

**Input Format:**
Linked list `1 -> 2 -> 2 -> 1`.

**Output Format:**
`True` if palindrome, else `False`.

**Example:**
```text
Input: 1 -> 2 -> 2 -> 1
Output: True
```



````python
#Given Linked list is a palindrome or not

class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        
def palindrome(head):
    if not head or not head.next:
        return True
    
    slow=fast=head
    while fast.next and fast.next.next:
        slow=slow.next
        fast=fast.next.next
        
    prev=None
    curr=slow.next
    while curr:
        nxt=curr.next
        curr.next=prev
        prev=curr
        curr=nxt
        
    first=head
    second=prev
    while second:
        if first.data!=second.data:
            return False
        first=first.next
        second=second.next
    return True

n=int(input())

if n==0:
    print(True)
else:
    arr=list(map(int,input().split()))
    
    head=Node(arr[0])
    temp=head
    for val in arr[1:]:
        temp.next=Node(val)
        temp=temp.next
        
    print(palindrome(head))
````

### Problem 17: Remove K-th Node from End of Linked List

**Problem Statement:**
Remove the $k$-th node from the end of a singly linked list using the two-pointer offset technique.

**Input Format:**
Linked list `1 -> 2 -> 3 -> 4 -> 5` and `k = 2`.

**Output Format:**
`1 -> 2 -> 3 -> 5`.

**Example:**
```text
Input: 1 2 3 4 5, k = 2
Output: 1 2 3 5
```



````python
#Removing kth element from the end of the linked list

class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        
def removeing(head,k):
    dummy=Node(0)
    dummy.next=head
    
    fast=slow=dummy
    
    for _ in range(k):
        fast=fast.next
        
    while fast.next:
        fast=fast.next
        slow=slow.next
    
    slow.next=slow.next.next
    
    return dummy.next

def display(head):
    temp=head
    while temp:
        print(temp.data,end=" -> ")
        temp=temp.next
    print("nullptr")
    
n=int(input())
arr=list(map(int,input().split()))
k=int(input())

head=Node(arr[0])
temp=head
for val in arr[1:]:
    temp.next=Node(val)
    temp=temp.next
    
head=removeing(head,k)
display(head)
````

### Problem 18: Linked List Cycle Detection

**Problem Statement:**
Detect if a linked list contains a cycle using Floyd's Tortoise and Hare algorithm.

**Input Format:**
Linked list with potential loop.

**Output Format:**
`True` if loop exists, else `False`.

**Example:**
```text
Input: 3 -> 2 -> 0 -> -4 (tail connects to node 1)
Output: True
```



````python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        
def merge_list(l1,l2):
    dummy=Node(0)
    tail=dummy
    
    while l1 and l2:
        if l1.data<l2.data:
            tail.next=l1
            l1=l1.next
        else:
            tail.next=l2
            l2=l2.next
        tail=tail.next

    if l1:
        tail.next=l1
    else:
        tail.next=l2
        
    return dummy.next
        
def display(head):
    temp=head
    while temp:
        print(temp.data,end=' -> ')
        temp=temp.next
    print("nullptr")
    
n1=int(input()) 
if n1>0:
    arr1=list(map(int,input().split()))
    head1=Node(arr1[0])
    temp=head1
    for val in arr1[1:]:
        temp.next=Node(val)
        temp=temp.next
else:
    head1=None
        
n2=int(input())
if n2>0:
    arr2=list(map(int,input().split()))
    head2=Node(arr2[0])
    temp=head2
    for val in arr2[1:]:
        temp.next=Node(val)
        temp=temp.next
else:
    head2=None
    
head=merge_list(head1,head2)
display(head)
````

### Problem 19: Asteroid Collision Simulation

**Problem Statement:**
Simulate the collision of asteroids moving in a line where positive moves right and negative moves left.

**Input Format:**
Array of asteroid sizes/directions.

**Output Format:**
State of remaining asteroids after all collisions.

**Example:**
```text
Input: [5, 10, -5]
Output: [5, 10]
```



````python
#astroids destroying (+/-) 

n=int(input())
asteroids=list(map(int,input().split()))

stack=[]

for ast in asteroids:
    destroyed=False
    
    while stack and ast<0 and stack[-1]>0:
        if abs(stack[-1])<abs(ast):
            stack.pop()
            continue
        elif abs(stack[-1])==abs(ast):
            stack.pop()
            destroyed=True
            break
        else:
            destroyed=True
            break
    if not destroyed:
        stack.append(ast)
        
if not stack:
    print("Everything Destroyed")
else:
    print(*stack)
````

### Problem 20: Top K Largest Numbers

The $k$ largest elements.

**Example:**
```text
Input: [3, 10, 4, 7, 2], k = 2

```



````python
#largest k no. of number

num=input()
k=int(input())

stack=[]

for digit in num:
    while stack and k>0 and stack[-1]>digit:
        stack.pop()
        k-=1
    stack.append(digit)
    
while k>0:
    stack.pop()
    k-=1
    
result=''.join(stack).lstrip('0')
print(result if result else "0")
````

### Problem 22: Sliding Window Maximum

**Problem Statement:**
Given an array and window size $k$, find the maximum element for each sliding window of size $k$.

**Input Format:**
Array and integer `k`.

**Output Format:**
List of window maximums.

**Example:**
```text
Input: [1, 3, -1, -3, 5, 3, 6, 7], k = 3
Output: [3, 3, 5, 5, 6, 7]
```



````python
#sliding window

from collections import deque

n=int(input())
arr=list(map(int,input().split()))
k=int(input())

dq=deque()
result=[]

for i in range(n):
    
    if dq and dq[0]<=i-k:
        dq.popleft()
    
    while dq and arr[dq[-1]]<arr[i]:
        dq.pop()
        
    dq.append(i)
    
    if i>=k-1:
        result.append(arr[dq[0]])
    
print(*result)
````

### Problem 23: Trapping Rain Water (Two Pointers)

**Problem Statement:**
Compute how much water can be trapped between elevation bars after raining using two pointers.

**Input Format:**
Line 1: Length `n`.
Line 2: `n` space-separated elevation heights.

**Output Format:**
Total units of trapped rainwater.

**Example:**
```text
Input:
6
0 1 0 2 1 0 1 3 2 1 2 1
Output: 6
```



````python
n=int(input())
height=list(map(int,input().split()))

left=0
right=n-1

leftmax=0
rightmax=0

water=0

while left<right:
    if height[left]<height[right]:
        if height[left]>=leftmax:
            leftmax=height[left]
        else:
            water+=leftmax-height[left]
        left+=1
    else:
        if height[right]>=rightmax:
            rightmax=height[right]
        else:
            water+=rightmax-height[right]
        right-=1

print(water)
````

### Problem 24: Postfix to Prefix Conversion

**Problem Statement:**
Convert a postfix arithmetic expression into prefix notation using an operand stack.

**Input Format:**
Postfix expression string.

**Output Format:**
Equivalent prefix expression.

**Example:**
```text
Input: AB+CD-*
Output: *+AB-CD
```



````python
def postfix_to_prefix(exp):
    stack=[]
    operators=['+','-','/','*']
    
    tokens=exp.split()
    
    for token in tokens:
        
        if token not in operators:
            stack.append(token)
        else:
            right=stack.pop()
            left=stack.pop()
            
            new_exp=token+" "+left+" "+right
            stack.append(new_exp)
            
    return stack[-1]

exp=input().strip()

print(postfix_to_prefix(exp))
````

### Problem 25: Postfix to Infix Conversion

**Problem Statement:**
Convert a postfix expression into fully parenthesized infix notation using a stack.

**Input Format:**
Postfix expression string.

**Output Format:**
Equivalent infix expression.

**Example:**
```text
Input: ab*c+
Output: ((a*b)+c)
```



````python
def postfix_to_infix(exp):
    
    stack=[]
    operators=['+','-','/','*']
    
    tokens=exp.split()
    
    for token in tokens:
        if token not in operators:
            stack.append(token)
        else:
            right=stack.pop()
            left=stack.pop()
            
            new_exp="("+left+" "+token+" "+right+")"
            stack.append(new_exp)
    return stack[-1]

exp=input().strip()

print(postfix_to_infix(exp))
````

```text
Output: abc*+
```



````python
def precedence(op):
    if op=='+' or op=='-':
        return 1
    if op=='/' or op=='*':
        return 2
    return 0

def tokenize(exp):
    tokens=[]
    i=0
    
    while i<len(exp):
        if exp[i].isdigit():
            num=""
            while i<len(exp) and exp[i].isdigit():
                num+=exp[i]
                i+=1
            tokens.append(num)
        else:
            tokens.append(exp[i])
            i+=1
    return tokens

def infix_to_prefix(exp):
    tokens=tokenize(exp)
    tokens=tokens[::-1]
    
    for i in range(len(tokens)):
        if tokens[i]=='(':
            tokens[i]=')'
        elif tokens[i]==')':
            tokens[i]='('
            
    stack=[]
    postfix=[]
    
    for token in tokens:
        if token.isdigit():
            postfix.append(token)
        elif token=='(':
            stack.append(token)
        elif token==')':
            while stack and stack[-1]!='(':
                postfix.append(stack.pop())
            if stack:
                stack.pop()
        else:
            while stack and precedence(stack[-1])>precedence(token):
                postfix.append(stack.pop())
            stack.append(token)
            
    while stack:
        postfix.append(stack.pop())
    
    return "".join(postfix[::-1])

n=input().strip()

print(infix_to_prefix(n))
````

### Problem 33: Linked List - Detect Loop (Floyd's Algorithm)

**Problem Statement:**
Detect if a loop exists in a linked list using two pointers (slow and fast).

**Input Format:**
List values and position to connect tail to.

**Output Format:**
`True` if loop exists, else `False`.

**Example:**
```text
Input: 1 2 3 4, loop to pos 1
Output: True
```



````python
class Node:
    def __init__(self, val):
        self.data = val
        self.next = None

def loopHere(head, tail, position):
    if position == 0:
        return

    walk = head
    for i in range(1, position):
        walk = walk.next

    tail.next = walk

def detectLoop(head):
    slow=head
    fast=head

    while fast is not None and fast.next is not None:
        slow=slow.next
        fast=fast.next.next

        if slow == fast:
            return True
    return False

n = int(input())

arr = list(map(int, input().split()))[:n]
num = arr[0]
head = tail = Node(num)

for i in range(n-1):
    tail.next = Node(arr[i+1])
    tail = tail.next

pos = int(input())
loopHere(head, tail, pos) 

ans = detectLoop(head)
if ans:
    print("True") 
else:
    print("False")
````

### Problem 35: Asteroid Collision (LeetCode 735 - Medium)

**Problem Statement:**
Given an array `asteroids` representing asteroids in a row, return the state of the asteroids after all collisions.

**Input Format:**
Asteroids array.

**Output Format:**
Remaining asteroids array.

**Example:**
```text
Input: [5, 10, -5]
Output: [5, 10]
```



````python
#735. Asteroid Collision (MEDIUM)

n=int(input())
asteroids=list(map(int,input().split()))

stack=[]

for ast in asteroids:
    destroyed=False
    
    while stack and ast<0 and stack[-1]>0:
        if abs(stack[-1])<abs(ast):
            stack.pop()
            continue
        elif abs(stack[-1])==abs(ast):
            stack.pop()
            destroyed=True
            break
        else:
            destroyed=True
            break
    if not destroyed:
        stack.append(ast)
        
if not stack:
    print("Everything Destroyed")
else:
    print(*stack)
````

### Problem 36: Trapping Rain Water (LeetCode 42 - Hard)

**Problem Statement:**
Given $n$ non-negative integers representing an elevation map where the width of each bar is 1, compute how much water it can trap after raining.

**Input Format:**
Elevation heights list.

**Output Format:**
Total trapped rainwater.

**Example:**
```text
Input: [0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]
Output: 6
```



````python
#42. Trapping Rain Water (HARD)

n=int(input())
height=list(map(int,input().split()))

left=0
right=n-1

leftmax=0
rightmax=0

water=0

while left<right:
    if height[left]<height[right]:
        if height[left]>=leftmax:
            leftmax=height[left]
        else:
            water+=leftmax-height[left]
        left+=1
    else:
        if height[right]>=rightmax:
            rightmax=height[right]
        else:
            water+=rightmax-height[right]
        right-=1

print(water)
````
